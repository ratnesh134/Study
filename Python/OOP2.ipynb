```python
class Point:

    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __str__(self):
        return '<{},{}>'.format(self.x,self.y)

    def euclidean_dist(self,other):
        return (((self.x - other.x)**2 + (self.y - other.y)**2)**0.5)

    def distance_from_origin(self):
        return self.euclidean_dist(Point(0,0))

class Line:

    def __init__(self,a,b,c):
        self.a = a
        self.b = b
        self.c = c

    def __str__(self):
        return '{}x + {}y + c = 0'.format(self.a,self.b,self.c)

    def point_on_line(line,point):
        if line.a * point.x + line.b * point.y + line.c == 0 :
            return "lies on the line"
        else:
            return "does not lie on the line"

    def shortest_distance_from_point(line,point):
        return abs(line.a * point.x + line.b * point.y + line.c)/(line.a**2 + line.b**2)**0.5
        
```

```python
p1 = Point(3,5)
p2 = Point(10,10)

p1.euclidean_dist(p2)
```

### Reference Variable

1. These variables hold the object
2. we can create objects without reference variable as well
3. An object can have multiple reference variables
4. Assigning a new reference variable to an existing object does not create a new object

```python
# Object without a reference

class Person:
    def __init__(self,name,country):
        self.name = name
        self.country = country


Person("Ratnesh","India")
```

The above method also creates the object and gives us the object's address

```python
# But if we do 

p = Person("Ratnesh","India")
```

This also creates the object , but p is technically not the object but it stores the object's address .

when we did Person() this command creates a object and then we are simply storing the object address and hence p is just a reference variable.

```python

# Now if we do 

q = p

# Aliasing
# Now q is also a reference variable , pointing to the same object location
```

```python
print(id(p))

print(id(q))
```

```python
# Chane attributes value with the help of 2nd object

print(p.name)
print(q.name)

q.name = "Happy"

print(p.name)
print(q.name)
```

```python
p.gender = "Male"
```

```python
q.gender
```

Creating an attribute outside the class will be shared between multiple reference variables as well.

## Pass by Reference

```python
class Person:

    def __init__(self,name,country):
        self.name = name
        self.country = country

# outside class ----> Function
def greet(person):
    print(f"Hello my name is {person.name} and I am from {person.country}")
```

```python
p = Person("Ratnesh","India")
# We can pass object as an input to the functions as well
x = greet(p)

```

```python
# We can do the reverse as well where we return an object from a class to another object

class Person:

    def __init__(self,name,country):
        self.name = name
        self.country = country

# outside class ----> Function
def greet(person):
    print(f"Hello my name is {person.name} and I am from {person.country}")
    p1 = Person("ankit","bangladesh")
    return p1

```

```python
p = Person("Ratnesh","India")
x = greet(p)

print(x.name)
print(x.country)
```

```python
# Excercise

class Person:

    def __init__(self,name,country):
        self.name = name
        self.country = country

# outside class ----> Function
def greet(person):
    print(id(person))
    print(f"Hello my name is {person.name} and I am from {person.country}")


p = Person("Ratnesh","India")
print(id(p))
greet(p)
```

In the above code snippet , the id of the object inside the function as well as outside the class would be same.

Because we are not passing the object , but object reference / object address to the function .

```python
# Excercise

class Person:

    def __init__(self,name,country):
        self.name = name
        self.country = country

# outside class ----> Function
def greet(person):
    print(id(person))
    print(f"Hello my name is {person.name} and I am from {person.country}")
    person.name = 'rahul'
    print(person.name)
    

p = Person("Ratnesh","India")
print(id(p))
greet(p)
print(p.name)
```

In the above code snippet , objects attributes can be changed through functions as well, that has been demonstrated above

## Objects mutability

```python
# Excercise

class Person:

    def __init__(self,name,country):
        self.name = name
        self.country = country

# outside class ----> Function
def greet(person):
    print(id(person))
    print(f"Hello my name is {person.name} and I am from {person.country}")
    person.name = 'rahul'
    return person
    

p = Person("Ratnesh","India")
print(id(p))
p1 = greet(p)
print(id(p1))
```

User defined classes objects are mutable . we can create immutable objects as well.

for immutable objects the address would be different
