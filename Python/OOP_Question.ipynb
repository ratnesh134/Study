```python
class Rectangle:

    def __init__(self,l,b):
        self.len = l
        self.width = b
        
        
    # Perimeter method
    def perimeter(self):
        peri = 2*(self.len + self.width)
        return peri

    # Area method
    def area(self):
        area = self.len * self.width
        return area

    # display method
    def display(self):
        print("length of the rectangle",self.len)
        print("Width of the rectangle",self.width)
        print("Perimeter of the rectangle is ",self.perimeter())
        print("Area of the rectangle is ",self.area())

    def __str__(self):
        return "Length of the rectangle is {}\nWidth of the rectangle is {}\nPerimeter of the rectangle is {}\nArea of the rectangle is {}".format(
        self.len,
        self.width,
        self.perimeter(),
        self.area()
    )

        
        
```

```python
rect = Rectangle(4,5)
```

```python
print(rect)
```

```python
class BankAccount:

    def __init__(self,acc_num=int,name=str,bal=int):
        self.acc_num = acc_num
        self.name = name
        self.bal = bal

    def deposit(self):
        dep_amt = int(input("Enter the amount to be deposited"))
        self.bal = self.bal+ dep_amt
        print("Amount deposited successfully,the current balance is ",self.bal)

    def withdrawal(self):
        withdraw_amt = int(input("Enter the value to withdraw from your account "))
        if withdraw_amt > self.bal:
            return "Insufficient Balance"

        else :
            self.bal = self.bal - withdraw_amt
            print("Amount withdrawn successfully, the current balance is ",self.bal)

    def bank_fees(self):
        fees_amt = (self.bal * 5)/100
        self.bal = self.bal - fees_amt
        print(f"The amount () is deducted as bank fees , the current balance is ()",fees_amt ,self.bal)

    def display(self):
        print("The current balance is ",self.bal)
        print("The account number is ",self.acc_num)
        print("Account holder name ",self.name)
    
```

```python
acc = BankAccount(23456789,"Ratnesh",2000)
```
