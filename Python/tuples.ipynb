# Tuples

A tuple in Python is similar to a list . The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, tuple is an immutable list . A tuple cannot be changed in any way once it is created.

Characteristics

1. Ordered
2. Unchangeable
3. Allows duplicates

```python
l = (1,2,2,3)

print(l)
```

```python
type(l)
```

```python
# empty tuple

t1 = ()
print(t1)

print(type(t1))
```

```python
# single item tuple

t2 = ("hello")
print(type(t2))
```

## Differences between Tuples and Lists

1. Syntax ---->  () for Tuples // [] for lists
2. Mutability ----> List are mutable ,can be changed meanwhile tuples cannot be changed.
3. Speed ----> tuples are faster than list . Generally , immutable datatypes are faster than mutable datatypes.
4. Memory ----> tuples take less memory in comparison to list.
5. Built-in-functionality ---> b-i-f are more in list in comparison to tuples
6. Error prone ---> list are more error prone than tuples.
7. Usabiity ----> if you want to make changes further then use lists , for read only operations use tuples

# Sets

A set is an unordered collections of items . Every set element is unique (no duplicates) and must be immutable (cannot be changed).


However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric differences etc.

Characteristics:
1. Unordered
2. Mutable
3. No duplicates
4. Can't contain mutable data types

Note - 2 D sets are not possible because of immutable elements, set within a set.

## Creating sets

```python
# empty set
# wrong method to create empty set
s = {}
print(s)
print(type(s))
```

```python
# Correct way
s = set()
print(s)
print(type(s))
```

```python
# 1 D sets
s = {1,2,3,4}
print(s)

# 2 D sets
s1 = {1,2,3,4,{2,4}}
```

therefore above we prove that 2D set is not possible .
Because it won't contain mutable datatypes as elements.

```python
# homo and hetro sets
s = {1,2.3,"hello",True,(1,2,3)}
print(s)
```

In the above example True element vanished because in Python True is treated as 1.

Also the order has also been changed because under the hood it uses hashing technique , therefore it is unordered.

```python
# Using type conversion
s = set([1,2,3,4,5])
print(s)
print(type(s))
```

```python
# Indexing , Slicing is not applicable to sets

s = {1,2,4,5,6}
s[0]
```

### Adding items to set

```python
s = {1,2,3,4,5}

# add 
s.add(8)
print(s)
```

```python
# update
# this is used to add multiple items
s.update([7,8,9])
print(s)
```

### Deleting items

```python
# del
# using this we can delete the entire set
s = {1,2,3,4}
print(s)
del s
print(s)
```

using del keyword we can delete the entire set, but not an individual item because indexing is not allowed in set.

```python
# discard
# here we can specify which element to delete from the set
s = {1,2,3,4,5,6}
print(s)
s.discard(6)
print(s)
```

if the element is not present in case of discard, then it won't throw an error.

But with remove function we will get an error

```python

# remove
s.remove(9)
```

```python
s.pop()
# deletes a random item from the set
print(s)
```

```python
s.clear()
# clears the elements from the set
# returns empty set
```

```python
print(s)
```

## Set operations

```python
s1 = {1,2,3,4,5,6}
s2 = {4,5,6,7,8,9,10}
# union
s1 | s2
```

```python
# Intersection
s1 & s2
```

```python
#Difference
s1 - s2
```

```python
# Symmetric Difference
# Gives the element which are not common in  both sets
s1 ^ s2
```

```python
# Membership test

1 not in s1
```

### set functions

```python
#len / min / max / sum / sorted

s = {1,2,4,5,6,7,8}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s,reverse = True))
```

```python
# union / update

# union is a temproary operation
s1 = {1,2,3,4,5,6}
s2 = {2,6,7,8,9}

print(s1.union(s2))
print(s1)
```

```python
# update is a permanent operation
s1.update(s2)   # or union_update

print(s1)
```

```python
# intersection and intersection_update
s1 = {1,2,3,4,5,6}
s2 = {2,6,7,8,9}
# temporary operation
print(s1.intersection(s2))

print(s1)
```

```python
# permanent operation is intersection_update

s1.intersection_update(s2)

print(s1)
```

```python
# isdisjoint 
s1 = {1,2,3,4,5}
s2 = {3,4,5,6,7}

s1.isdisjoint(s2)
```

```python
# issubset function
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s2.issubset(s1)
```

```python
# issuperset
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issuperset(s2)
```

```python
# copy

s1 = {1,2,3,4,5}
s2 = s1.copy() 
# creates a new set with the same elements. 
# Creates a shallow copy

print(s1)
print(s2)

print(id(s1))
print(id(s2))
# therefore their Ids will also be different if it was s1 = s2
# then in that case both will point to the same memory location.
```

## Frozenset

Frozen set is just an immutable version of a Python set object

Which means we won't be able to add new items to the set or delete any item from the set

What works ? ---> all read function

What doesn't works ? all write operation

```python
# Create frozen set

fs = frozenset([1,2,3,4])

fs
```

```python
fs.add(7)
```

```python
fs1 = frozenset([1,2,3,4])
fs2 = frozenset([5,6,7,8,9])

fs1.intersection(fs2)
```

```python
# 2 D frozenset

fs = frozenset([1,2,3,4,frozenset([7,8])])

fs
```
