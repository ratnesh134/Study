# What are functions ?
In Python, a function is a reusable block of code that performs a specific task.
 

Syntax for a function -




Functions core philosophy are 
1. Abstraction - Abstraction means hiding unnecessary implementation details and showing only what is needed.
2. Decomposition - Decomposition means breaking a large/complex problem into smaller, manageable functions.
3. Reusability
4. Modularity
5. Maintainability

```python
# Creating a function

def is_even(num):
    '''
    This function returns if a given number is odd or even
    input - any valid integer
    output - odd/even
    '''
    if num%2 == 0:
        return 'even'
    else:
        return 'odd'
```

```python
# Function calling

is_even(67)
```

## 2 Point of Views

```python

def is_even(num):
    '''
    This function returns if a given number is odd or even
    input - any valid integer
    output - odd/even
    '''

    if type(num) == int:
        if num%2 == 0:
            return 'even'
        else:
            return 'odd'
    else:

        return "Invalid Input"
```

```python
is_even("Hello")
```

## Parameters VS Arguments

Important Note - Whenever we create a function , then that variable inside the parenthesis is called "parameter". 

Now when we call the function then the value passed inside the function will be called as "arguments"

### Types of Arguments

1. Default Arguments
2. Positional Arguments
3. Keyword Arguments

```python
# Default Argument

def power(a=1,b=1):
    return a**b
    
```

```python
power(2,3)
```

```python
power(2)
```

```python
# Positional Arguments

power(2,3)
```

```python
#Keyword Arguments

power(b=3,a=2)
```

## *args AND **kwargs

These are special Python Keywords that are used to pass the variable length of arguments to a function.


*args ---> Allows us to pass a variable number of non-keyword arguments to a function. 

This would be helpful when the we don't know how many variable the user might pass to a function

Internally it stores the value into a tuple.

```python
# To mutiply n number of values passed by the user

def multiply(*args):
    '''
    This function takes a number of values passed
    by the user and mutiply them and give back the product
    '''
    product = 1
    for i in args:
        product = product * i

    return product
    
```

```python
multiply(1,2,3,4,5,6,7)
```

```python
# To get the documentation of any function

print(multiply.__doc__)
```

# **kwargs

**kwargs allows us to pass any number of keyword arguments.

Keyword argument means that they contain a key-value pair,like a Python Dictionary.

Internally here it stores the value into a python dictionary

```python
def display(**kwargs):
    for key,value in kwargs.items():
        print(key,'-->',value)
```

```python
display(India='New Delhi',USA='Washighton DC',nepal='Kathmandu')
```

### Note

Points to remember while using *args and **kwargs

1. Order of the arguments matters (normal arg,--> *args --> **kwargs) , this placement is necessary when we are dealing with a function where we are using all the three types of arguments

2.  The words 'args' and 'kwwargs' are only a convention we can use any name of our choice.
