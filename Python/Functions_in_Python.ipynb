# What are functions ?
In Python, a function is a reusable block of code that performs a specific task.
 

Syntax for a function -




Functions core philosophy are 
1. Abstraction - Abstraction means hiding unnecessary implementation details and showing only what is needed.
2. Decomposition - Decomposition means breaking a large/complex problem into smaller, manageable functions.
3. Reusability
4. Modularity
5. Maintainability

```python
# Creating a function

def is_even(num):
    '''
    This function returns if a given number is odd or even
    input - any valid integer
    output - odd/even
    '''
    if num%2 == 0:
        return 'even'
    else:
        return 'odd'
```

```python
# Function calling

is_even(67)
```

## 2 Point of Views

```python

def is_even(num):
    '''
    This function returns if a given number is odd or even
    input - any valid integer
    output - odd/even
    '''

    if type(num) == int:
        if num%2 == 0:
            return 'even'
        else:
            return 'odd'
    else:

        return "Invalid Input"
```

```python
is_even("Hello")
```
