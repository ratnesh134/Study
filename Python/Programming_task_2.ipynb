```python
# find the maximum sub-array in a list

def max_sub_array(l):

    max_val = sum(l)
    max_subarray = l

    for i in range(0,len(l)):
        sub_array = []
        for j in range(i,len(l)):
            sub_array.append(l[j])
            if max_val <= sum(sub_array):
                max_val = sum(sub_array)
                max_subarray = sub_array

            else:
                continue
    return max_val,max_subarray


l = [-2,4,7,-1,6,-11,14,3,-1,-6]
a,b = max_sub_array(l)

print(f'the maximum sub array is {b} with value {a}')
            
```

### What is aliasing ?

Aliasing means that two or more variables refer to the same object in memory.



```python
a = [1, 2, 3]
b = a
```

```python
#         ┌───────────────┐
# a ─────►│  [1, 2, 3]    │
#         └───────────────┘
#              ▲
#              │
# b ───────────┘
```

Both a and b refer to the same list object.


You can verify this using id()

```python
id(a)
```

```python
id(b)
```

```python
id([1,2,3])
```

```python
a is [1,2,3]
```

```python
x = 4
y = x
```

```python
print(id(x))
print(id(y))
print(id(4))
```

```python
x is 4
```

The key difference

Compare these two:

a = a + 1

and

a.append(1)

The first reassigns a to another object.

The second modifies the existing object.

A useful mental model

Always ask yourself two questions:

1. Is the object mutable or immutable?

2. Am I modifying the object, or am I making the variable refer to another object?
