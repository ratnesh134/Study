### What is aliasing ?

Aliasing means that two or more variables refer to the same object in memory.

Aliasing is a technique by which we optimize memory in Python by creating a reference .

### Mutable Datatype

the datatypes in which we can change the values at the same memory location

```python
# for example

l = [1,2,3]
print(id(l))
l.append(5)
print(l)
print(id(l))
```

### Immutable Datatype

The datatype where we cannot change the value once declared, at the same memory location.

```python
# for example

t = (1,2,4)

print(id(t))

t = t + (5,)
print(t)
print(id(t))
```

#### Dangerous Scenarios


```python
l = [1,2,3,4]
m = l
m.append(8)
print(m)
print(l)
```

Here both the variables/ names are getting updated because they are pointing towards the same object.So using them in programs can be risky at time.

### Difference between deep and shallow copy ?

#### Shallow copy

Creates a new container but copies references to the objects contained inside.

#### Deep copy

Recursively creates copies of contained objects, subject to Python's copying rules.

So shallow copying doesn't mean "nothing inside is copied."

It means:

Outer object → copied

Objects referenced by outer object → references copied

Deep copying:

Outer object → copied

Nested mutable objects → recursively copied


```python
# Shallow Copy works for 1 dimensional list

l = [3,4,5]

m = l.copy()

print(id(l))
print(id(m))
```

```python
# Now if there is a 2 or more than 2 dimensional list it won't work the same

l = [[1,2],[4,5]]

m = l.copy()

print(id(l))
print(id(m))
```
