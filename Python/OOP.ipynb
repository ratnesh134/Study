```python
# ATM Application
# class name is written in Pascal Case for ex HelloWorld

class Atm:

    # Constructor ===> Special Function
    def __init__(self):
        self.pin = ''
        self.balance = 0
        self.menu()


    def menu(self):
        user_input = input("""
        Hi, How can I help you ?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)

        if user_input == '1':
            #create pin
            self.create_pin()
        if user_input == '2':
            # change pin
            self.change_pin()
        if user_input == '3':
            #check balance
            self.check_balance()
        if user_input == '4':
            #withdraw
            self.withdraw_balance()
        else :
            pass

    def create_pin(self):
        user_pin = input("Enter your pin : ")
        self.pin = user_pin

        user_balance = int(input("Enter Balance :"))
        self.balance = user_balance

        print("Pin Created Successfully")
        self.menu()

    def change_pin(self):
        old_pin = input("Enter existing pin : ")
        if old_pin == self.pin:
            new_pin = input("Enter new Pin :")
            self.pin = new_pin
            print("Pin Changes Successfully")
        else:
            print("Existing pin doesn't match, Can't change the pin")
        self.menu()

    def check_balance(self):
        user_pin = input("Enter your pin :")
        if user_pin == self.pin:
            print("Your current balance is ",self.balance)
        else:
            print("Entered Pin Is Incorrect")

        self.menu()

    def withdraw_balance(self):

        user_pin = input("Enter the pin :")
        if user_pin == self.pin:
            amt = int(input("Enter the amount to withdraw :"))
            if self.balance >= amt:
                rem_balance = self.balance - amt
                print("Amout withdrawal successful , your remaining balance is ",rem_balance)
            else:
                print("You don't have enough balance")
        else:
            print("Enter the correct pin")
        self.menu()
            
    
    
           
```

## Creating Datatype using OOP 

```python
# to create a datatype for fractions

class Fraction:

    # parameterized constructor
    def __init__(self,x,y):
        self.num = x
        self.den = y

    # 2nd magic method
    def __str__(self):
        return '{}/{}'.format(self.num,self.den)


    # 3rd magic method
    def __add__(self,other):
        new_num = self.num * other.den + self.den * other.num
        new_den = self.den * other.den

        return '{}/{}'.format(new_num,new_den)
```

```python
fra = Fraction(3,4)
```

```python
fra1 = Fraction(3,4)
fra2 = Fraction(4,5)
```

```python
id(fra)
```

```python
print(fra1 + fra2)
```
